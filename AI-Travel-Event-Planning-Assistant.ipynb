```python
import ipywidgets as widgets
from IPython.display import display, clear_output

title = widgets.HTML(
    "<h1>✈️ AI Travel & Event Planner</h1>"
    "<p>Plan your trip or event easily!</p>"
)

name_box = widgets.Text(
    description="Name:",
    placeholder="Enter your name"
)

type_box = widgets.Dropdown(
    options=["Travel", "Event"],
    description="Plan Type:"
)

destination_box = widgets.Text(
    description="Place:",
    placeholder="Enter destination or event"
)

days_box = widgets.IntSlider(
    value=3,
    min=1,
    max=10,
    description="Days:"
)

budget_box = widgets.IntText(
    value=5000,
    description="Budget ₹:"
)

button = widgets.Button(
    description="Generate Plan",
    button_style="success"
)

output = widgets.Output()

def generate_plan(b):
    with output:
        clear_output()

        name = name_box.value
        plan = type_box.value
        place = destination_box.value
        days = days_box.value
        budget = budget_box.value

        if not place.strip():
            print("Please enter a destination or event location.")
            return

        if budget <= 0:
            print("Please enter a valid budget.")
            return

        print("🌟 YOUR PERSONAL PLAN 🌟")
        print("Name:", name)
        print("Type:", plan)
        print("Location:", place)
        print("Duration:", days, "days")
        print("Budget: ₹", budget)

        if plan == "Travel":
            print("\n📍 DAY-WISE PLAN")

            if place.lower() == "ooty":
                activities = [
                    "Ooty Lake and Botanical Garden",
                    "Doddabetta Peak and Rose Garden",
                    "Tea Factory and Pykara Falls"
                ]
            else:
                activities = [
                    "Explore popular local attractions",
                    "Visit famous sightseeing spots",
                    "Enjoy local food and shopping"
                ]

            for i in range(days):
                print(
                    f"Day {i+1}: "
                    f"{activities[i % len(activities)]}"
                )

            print("\n💰 SUGGESTED BUDGET")
            print("Travel: ₹", round(budget * 0.30))
            print("Hotel: ₹", round(budget * 0.30))
            print("Food: ₹", round(budget * 0.25))
            print("Activities: ₹", round(budget * 0.15))

        else:
            print("\n🎉 EVENT CHECKLIST")
            print("1. Choose the venue")
            print("2. Prepare the guest list")
            print("3. Arrange food")
            print("4. Plan decorations")
            print("5. Prepare the event schedule")

            print("\n💰 EVENT BUDGET")
            print("Venue: ₹", round(budget * 0.30))
            print("Food: ₹", round(budget * 0.35))
            print("Decorations: ₹", round(budget * 0.20))
            print("Other: ₹", round(budget * 0.15))

button.on_click(generate_plan)

display(title)
display(name_box)
display(type_box)
display(destination_box)
display(days_box)
display(budget_box)
display(button)
display(output)
```
